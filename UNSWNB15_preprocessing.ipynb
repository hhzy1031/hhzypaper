### 字符型处理

```python
A.shape
```

```python
for i,j in zip(A.columns,A.dtypes):
    print(i,j)
```

```python
A[1] = A[1].apply(lambda x:str(x).replace('0x000b','').replace('0','')) # 将垂直制表符替换成0
A[39] = A[39].apply(lambda x:str(x).replace('0x000b','').replace('0','')) # 将垂直制表符替换成0
```

```python
A[1] = LabelEncoder().fit_transform(A[1]) # 对第一列编码
A[39] = LabelEncoder().fit_transform(A[39]) # 对第39列编码
```

```python
#将object列转换为float
A[1] = A[1].astype(float)
print(A[2])
A[2] = A[2].astype(float)
```

# 标准化

```python
from sklearn.preprocessing import StandardScaler

for x in range(A.shape[1]):
    d=A.iloc[:,x]
    try:
        A.iloc[:,x]=d.astype(float)
    except:
        A.iloc[:,x]=LabelEncoder().fit_transform(d)
    #print(x)  #输出特征序号
Dt1=StandardScaler().fit_transform(A)
```

```python
#将15分类标签数值化
i=0
for i in range(L10.shape[0]):
    if L10.iloc[i] == "Exploits":
        L10.iloc[i]=1
    elif (L10.iloc[i] == "Reconnaissance")|(L10.iloc[i] == " Reconnaissance "):
        L10.iloc[i]=2
    elif L10.iloc[i] == "DoS":
        L10.iloc[i]=3
    elif L10.iloc[i] == "Generic":
        L10.iloc[i]=4
    elif (L10.iloc[i] == "Shellcode")|(L10.iloc[i] == " Shellcode "):
        L10.iloc[i]=5
    elif (L10.iloc[i] == " Fuzzers")|(L10.iloc[i] == " Fuzzers "):
        L10.iloc[i]=6
    elif L10.iloc[i] == "Worms":
        L10.iloc[i]=7
    elif (L10.iloc[i] == "Backdoors")|(L10.iloc[i] == "Backdoor"):
        L10.iloc[i]=8
    elif L10.iloc[i] == "Analysis":
        L10.iloc[i]=9
        
        i=i+1
```
